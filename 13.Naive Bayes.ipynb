# movie review naive bayes

```python
df=pd.read_csv("IMDB Dataset.csv")
df.head(5)
```

```python
value = df.loc[0,'review']
print(value)
```

```python
df=df.sample(10000)  # we select 10000 from 50000 to reduce time
df.shape
```

```python
df['sentiment'].replace({'positive':1, 'negative':0},inplace=True)
```

```python
df.head(3)
```

```python
import re
clean = re.compile('<.*?>')
re.sub(clean,'',df.iloc[2].review)
```

```python
# Function to clean html tags
def clean_html(text):
    clean = re.compile(' <.*? >')
    return re.sub(clean,' ', text)
```

```python
df['review']=df['review'].apply(clean_html)
```

```python
def convert_lower(text):
    return text.lower()
```

```python
df['review']=df['review'].apply(convert_lower)
```

```python
# function to remove special characters

def remove_special(text):
    x=' '
    for i in text:
        if i.isalnum():
            x=x+i
        else:
            x=x + ' '

    return x
```

```python
df['review']=df['review'].apply(remove_special)
```

```python
import nltk
from nltk.corpus import stopwords

# Download the stopwords dataset (only needed once)
nltk.download('stopwords')

# Access the list of English stop words
print(stopwords.words('english'))
```

```python
df
```

```python
# 1. Convert stop words to a set ONCE globally (huge performance speedup)
STOPWORDS = set(stopwords.words('english'))

# 2. Optimized function using list comprehension and string join
def remove_stopwords(text):
    return " ".join([word for word in text.split() if word.lower() not in STOPWORDS])

# 3. Apply to dataframe
df['review'] = df['review'].apply(remove_stopwords)
```

```python
# Perform stemming

from nltk.stem.porter import PorterStemmer
ps=PorterStemmer()
```

```python
y=[ ]
def stem_words(text):
    for i in text:
        y.append(ps.stem(i))
    z=y[:]
    y.clear()
    return z
```

```python
stem_words(['I','loved','loving','it'])
```

```python
df['review']=df['review'].apply(stem_words)
```
